```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD ALL DATASETS
# ==========================================

panel_df = pd.read_csv('/kaggle/input/refined-asian-economic-indicators-complete-panel-dataset/Refined_Asian_Economic_Indicators_Complete_Panel_Dataset.csv')

annual_df = pd.read_csv('/kaggle/input/asian-economic-indicators-annual-summary/Asian_Economic_Indicators_Annual_Summary.csv')

country_df = pd.read_csv('/kaggle/input/asian-economic-indicators-country-coverage-summary/Asian_Economic_Indicators_Country_Coverage_Summary.csv')

print("Panel Dataset Shape:", panel_df.shape)
print("Annual Summary Shape:", annual_df.shape)
print("Country Coverage Shape:", country_df.shape)
```

```

# ==========================================
# PREVIEW DATASETS
# ==========================================

panel_df.head()
```

```

annual_df.head()
```

```

country_df.head()
```


# 📊 Panel Dataset Analysis


```

panel_df.info()
```

```

numeric_cols = panel_df.select_dtypes(include='number').columns.tolist()

country_col = [c for c in panel_df.columns if 'country' in c.lower()][0]
year_col = [c for c in panel_df.columns if 'year' in c.lower()][0]

main_metric = numeric_cols[0]

print("Main Metric:", main_metric)
```

```

# ==========================================
# DISTRIBUTION ANALYSIS
# ==========================================

fig = px.histogram(
    panel_df,
    x=main_metric,
    nbins=50,
    template='plotly',
    title=f'Distribution of {main_metric}'
)

fig.show(renderer='iframe_connected')
```

```

# ==========================================
# CORRELATION HEATMAP
# ==========================================

corr = panel_df[numeric_cols[:8]].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Economic Indicators Correlation Heatmap')

plt.show()
```


# 🌍 Top Economies


```

top_country = (
    panel_df.groupby(country_col)[main_metric]
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_country,
    x=country_col,
    y=main_metric,
    color=main_metric,
    template='plotly',
    title='Top Economies Comparison'
)

fig.show(renderer='iframe_connected')
```


# 📈 Economic Trend Analysis


```

top5 = (
    panel_df.groupby(country_col)[main_metric]
    .mean()
    .sort_values(ascending=False)
    .head(5)
    .index
)

trend = panel_df[panel_df[country_col].isin(top5)]

fig = px.line(
    trend,
    x=year_col,
    y=main_metric,
    color=country_col,
    template='plotly',
    title='Top Countries Economic Trend'
)

fig.show(renderer='iframe_connected')
```


# 🗺️ Economic Performance Map


```

country_avg = (
    panel_df.groupby(country_col)[main_metric]
    .mean()
    .reset_index()
)

fig = px.choropleth(
    country_avg,
    locations=country_col,
    locationmode='country names',
    color=main_metric,
    hover_name=country_col,
    template='plotly',
    title='Asian Economic Performance Map'
)

fig.show(renderer='iframe_connected')
```


# 📅 Annual Summary Dataset Analysis


```

annual_df.info()
```

```

annual_numeric = annual_df.select_dtypes(include='number').columns.tolist()

annual_metric = annual_numeric[0]

fig = px.line(
    annual_df,
    x=annual_df.columns[0],
    y=annual_metric,
    markers=True,
    template='plotly',
    title='Annual Economic Summary Trend'
)

fig.show(renderer='iframe_connected')
```


# 🌏 Country Coverage Analysis


```

country_df.info()
```

```

country_numeric = country_df.select_dtypes(include='number').columns.tolist()

coverage_metric = country_numeric[0]

fig = px.bar(
    country_df,
    x=country_df.columns[0],
    y=coverage_metric,
    color=coverage_metric,
    template='plotly',
    title='Country Coverage Comparison'
)

fig.show(renderer='iframe_connected')
```


# 🚀 Ultimate Interactive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Top Economies',
        'Trend Analysis',
        'Coverage Analysis'
    )
)

# Histogram
fig.add_trace(
    go.Histogram(
        x=panel_df[main_metric]
    ),
    row=1,
    col=1
)

# Top countries
top10 = (
    panel_df.groupby(country_col)[main_metric]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)

# Trend
trend2 = (
    panel_df.groupby(year_col)[main_metric]
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=trend2.index,
        y=trend2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)

# Coverage
coverage = (
    country_df.groupby(country_df.columns[0])[coverage_metric]
    .mean()
)

fig.add_trace(
    go.Bar(
        x=coverage.index,
        y=coverage.values
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=950,
    template='plotly',
    title='🌏 Asia Economic Intelligence Dashboard',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
