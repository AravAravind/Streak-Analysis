# 🌏 Asia Economic Intelligence Dashboard

Premium Kaggle-ready notebook with interactive visualizations and storytelling.

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio

pio.renderers.default = "iframe_connected"

sns.set_style("darkgrid")

print("Libraries loaded!")
```

```

df = pd.read_csv('/kaggle/input/refined-asian-economic-indicators-complete-panel-dataset/Refined_Asian_Economic_Indicators_Complete_Panel_Dataset.csv')

print(df.shape)

df.head()
```

```

df.info()
```

```

numeric_cols = df.select_dtypes(include='number').columns.tolist()

country_col = [c for c in df.columns if 'country' in c.lower()][0]
year_col = [c for c in df.columns if 'year' in c.lower()][0]

main_metric = numeric_cols[0]
```

```

fig = px.histogram(
    df,
    x=main_metric,
    nbins=50,
    template='plotly',
    title=f'Distribution of {main_metric}'
)

fig.show(renderer='iframe_connected')
```

```

corr = df[numeric_cols[:8]].corr(numeric_only=True)

plt.figure(figsize=(12,7))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title('Economic Correlation Heatmap')

plt.show()
```

```

top_country = (
    df.groupby(country_col)[main_metric]
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_country,
    x=country_col,
    y=main_metric,
    color=main_metric,
    template='plotly',
    title='Top Economies'
)

fig.show(renderer='iframe_connected')
```

```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Top Economies',
        'Trend',
        'Scatter'
    )
)

fig.add_trace(
    go.Histogram(x=df[main_metric]),
    row=1,
    col=1
)

top10 = (
    df.groupby(country_col)[main_metric]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)

trend = (
    df.groupby(year_col)[main_metric]
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=trend.index,
        y=trend.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=df[numeric_cols[0]],
        y=df[numeric_cols[1]],
        mode='markers'
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=900,
    template='plotly',
    title='🌏 Asia Economic Dashboard',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
