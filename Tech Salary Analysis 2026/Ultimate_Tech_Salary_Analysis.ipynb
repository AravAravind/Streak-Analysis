```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries imported successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(r"/mnt/data/tech_salary_dataset.csv")

print("Dataset Loaded Successfully!")
print(df.shape)
```

```

# ==========================================
# PREVIEW DATA
# ==========================================

df.head()
```

```

# ==========================================
# COLUMN NAMES
# ==========================================

df.columns.tolist()
```


# 🧹 Data Cleaning


```

# Missing Values
df.isnull().sum()
```

```

# Remove duplicates
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)

df = df.drop_duplicates()

print("New Shape:", df.shape)
```

```

# Basic Info
df.info()
```

```

# Descriptive Statistics
df.describe(include='all')
```


# 📊 Exploratory Data Analysis


```

# ==========================================
# TARGET DISTRIBUTION
# ==========================================

fig = px.histogram(
    df,
    x='job_satisfaction_score',
    nbins=50,
    title='Distribution of job_satisfaction_score',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```

```

# ==========================================
# TOP CATEGORIES
# ==========================================

top_values = df['employee_id'].value_counts().head(15)

fig = px.bar(
    x=top_values.values,
    y=top_values.index,
    orientation='h',
    title='Top employee_id Categories',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```

```

# ==========================================
# CATEGORY VS SALARY
# ==========================================

avg_salary = (
    df.groupby('employee_id')['job_satisfaction_score']
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    avg_salary,
    x='employee_id',
    y='job_satisfaction_score',
    color='job_satisfaction_score',
    title='Average Salary by employee_id',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```

```

# ==========================================
# CORRELATION HEATMAP
# ==========================================

corr = df[['age', 'experience_years_total', 'experience_years_in_field', 'num_skills', 'annual_salary_usd', 'monthly_salary_usd', 'hourly_rate_usd', 'annual_bonus_usd']].corr(numeric_only=True)

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```

# ==========================================
# OUTLIER ANALYSIS
# ==========================================

fig = px.box(
    df,
    y='job_satisfaction_score',
    title='Salary Outlier Detection',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🌍 Interactive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    specs=[
        [{"type": "xy"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "xy"}]
    ],
    subplot_titles=(
        'Distribution',
        'Top Categories',
        'Outliers',
        'Summary'
    )
)


fig.add_trace(
    go.Histogram(x=df['job_satisfaction_score']),
    row=1,
    col=1
)

fig.add_trace(
    go.Box(y=df['job_satisfaction_score']),
    row=2,
    col=1
)

top_values = df['employee_id'].value_counts().head(10)

fig.add_trace(
    go.Bar(
        x=top_values.index,
        y=top_values.values
    ),
    row=1,
    col=2
)

summary = (
    df.groupby('employee_id')['job_satisfaction_score']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=summary.index,
        y=summary.values
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=900,
    title='🚀 Interactive Tech Salary Dashboard',
    template='plotly',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
