```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
import plotly.offline as pyo

pio.renderers.default = "notebook_connected"
pyo.init_notebook_mode()

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv('/kaggle/input/your-dataset/submission.csv')

print("Dataset Loaded Successfully!")

print(df.shape)

df.head()
```

```

# ==========================================
# DATASET OVERVIEW
# ==========================================

df.info()
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 📈 Prediction Distribution Analysis


```

fig = px.histogram(
    df,
    x='PitNextLap',
    nbins=60,
    template='plotly_dark',
    title='Prediction Distribution'
)

fig.show()
```


# 📦 Prediction Spread & Outliers


```

fig = px.box(
    df,
    y='PitNextLap',
    template='plotly_dark',
    title='Prediction Outlier Analysis'
)

fig.show()
```


# 🔥 Density Curve


```

fig = px.violin(
    df,
    y='PitNextLap',
    box=True,
    points='all',
    template='plotly_dark',
    title='Prediction Density Analysis'
)

fig.show()
```


# 🏆 Top Predictions


```

top_preds = (
    df.sort_values('PitNextLap', ascending=False)
    .head(20)
)

fig = px.bar(
    top_preds,
    x='id',
    y='PitNextLap',
    color='PitNextLap',
    template='plotly_dark',
    title='Top Prediction Scores'
)

fig.show()
```


# 📊 Percentile Intelligence


```

df['percentile_rank'] = (
    df['PitNextLap']
    .rank(pct=True)
)

fig = px.scatter(
    df,
    x=df.index,
    y='percentile_rank',
    color='PitNextLap',
    template='plotly_dark',
    title='Prediction Percentile Ranking'
)

fig.show()
```


# 🚀 Executive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Prediction Distribution',
        'Outlier Analysis',
        'Top Predictions',
        'Percentile Ranking'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['PitNextLap']
    ),
    row=1,
    col=1
)

fig.add_trace(
    go.Box(
        y=df['PitNextLap']
    ),
    row=1,
    col=2
)


top10 = (
    df.sort_values('PitNextLap', ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10['id'],
        y=top10['PitNextLap']
    ),
    row=2,
    col=1
)


df['percentile_rank'] = df['PitNextLap'].rank(pct=True)

fig.add_trace(
    go.Scatter(
        x=df.index,
        y=df['percentile_rank'],
        mode='markers'
    ),
    row=2,
    col=2
)


fig.update_layout(
    height=950,
    template='plotly_dark',
    title='🚀 Submission Intelligence Dashboard',
    showlegend=False
)

fig.show()
```
