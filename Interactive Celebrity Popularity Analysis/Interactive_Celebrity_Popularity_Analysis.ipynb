```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go

from plotly.subplots import make_subplots

import warnings
warnings.filterwarnings('ignore')

# Plot settings
plt.rcParams['figure.figsize'] = (12,6)
sns.set_style('darkgrid')

print("Libraries imported successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv('/kaggle/input/popular-people/popular_people.csv')

print("Dataset Loaded Successfully!")
```

```

# ==========================================
# FIRST LOOK
# ==========================================

df.head()
```

```

# ==========================================
# DATASET SHAPE
# ==========================================

rows, cols = df.shape

print(f"Rows    : {rows}")
print(f"Columns : {cols}")
```

```

# ==========================================
# DATASET INFO
# ==========================================

df.info()
```


# 🧹 Data Cleaning

Before visualization, let's clean the dataset carefully.


```

# ==========================================
# CHECK MISSING VALUES
# ==========================================

missing = df.isnull().sum()

missing
```

```

# ==========================================
# HANDLE MISSING VALUES
# ==========================================

df['known_for_department'] = df['known_for_department'].fillna('Unknown')

print("Missing values handled!")
```

```

# ==========================================
# CHECK DUPLICATES
# ==========================================

duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```

# ==========================================
# REMOVE DUPLICATES
# ==========================================

df = df.drop_duplicates()

print("Duplicates removed!")
print(df.shape)
```


# 🎭 Gender Mapping

The dataset uses encoded values for gender.

| Code | Gender |
|---|---|
| 0 | Unknown |
| 1 | Female |
| 2 | Male |
| 3 | Non-Binary |


```

# ==========================================
# MAP GENDER VALUES
# ==========================================

gender_map = {
    0: 'Unknown',
    1: 'Female',
    2: 'Male',
    3: 'Non-Binary'
}

df['gender_label'] = df['gender'].map(gender_map)

df[['gender', 'gender_label']].head()
```


# 📊 Descriptive Statistics


```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 🎨 Univariate Analysis


```

# ==========================================
# POPULARITY DISTRIBUTION
# ==========================================

fig = px.histogram(
    df,
    x='popularity',
    nbins=60,
    title='Popularity Distribution',
    template='plotly_dark'
)

fig.show()
```


### 🔍 Insight

The popularity distribution is highly right-skewed.

This means:
- Most celebrities have low-to-medium popularity
- A small number dominate the industry


```

# ==========================================
# LOG TRANSFORMATION
# ==========================================

df['log_popularity'] = np.log1p(df['popularity'])

fig = px.histogram(
    df,
    x='log_popularity',
    nbins=60,
    title='Log Transformed Popularity Distribution',
    template='plotly_dark'
)

fig.show()
```

```

# ==========================================
# GENDER DISTRIBUTION
# ==========================================

gender_counts = df['gender_label'].value_counts()

fig = px.pie(
    values=gender_counts.values,
    names=gender_counts.index,
    title='Gender Distribution',
    hole=0.5,
    template='plotly_dark'
)

fig.show()
```

```

# ==========================================
# DEPARTMENT DISTRIBUTION
# ==========================================

dept = df['known_for_department'].value_counts()

fig = px.bar(
    x=dept.values,
    y=dept.index,
    orientation='h',
    title='Department Distribution',
    labels={'x':'Count', 'y':'Department'},
    template='plotly_dark'
)

fig.show()
```


# 🌟 Top Celebrities


```

# ==========================================
# TOP 20 MOST POPULAR PEOPLE
# ==========================================

top20 = df.sort_values('popularity', ascending=False).head(20)

fig = px.bar(
    top20,
    x='popularity',
    y='name',
    orientation='h',
    color='popularity',
    title='Top 20 Most Popular Celebrities',
    template='plotly_dark'
)

fig.show()
```


# 📈 Department vs Popularity


```

# ==========================================
# BOXPLOT
# ==========================================

fig = px.box(
    df,
    x='known_for_department',
    y='popularity',
    color='known_for_department',
    title='Popularity by Department',
    template='plotly_dark'
)

fig.show()
```


### 🔍 Insight

Actors dominate both:
- Quantity
- Popularity scores

Some departments contain extreme outliers.


```

# ==========================================
# VIOLIN PLOT
# ==========================================

fig = px.violin(
    df,
    x='known_for_department',
    y='popularity',
    color='known_for_department',
    box=True,
    title='Popularity Density by Department',
    template='plotly_dark'
)

fig.show()
```


# 🌳 Treemap Visualization


```

# ==========================================
# TREEMAP
# ==========================================

fig = px.treemap(
    df,
    path=['known_for_department'],
    values='popularity',
    color='popularity',
    title='Popularity Treemap',
    template='plotly_dark'
)

fig.show()
```


# ☀️ Sunburst Chart


```

# ==========================================
# SUNBURST CHART
# ==========================================

fig = px.sunburst(
    df,
    path=['known_for_department', 'gender_label'],
    values='popularity',
    title='Department & Gender Sunburst',
    template='plotly_dark'
)

fig.show()
```


# 📌 Correlation Analysis


```

# ==========================================
# CORRELATION HEATMAP
# ==========================================

corr = df[['gender', 'popularity']].corr()

plt.figure(figsize=(8,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```


# 🚨 Outlier Detection


```

# ==========================================
# OUTLIER VISUALIZATION
# ==========================================

fig = px.box(
    df,
    y='popularity',
    title='Popularity Outliers',
    template='plotly_dark'
)

fig.show()
```


# 🏆 Most Dominant Departments


```

# ==========================================
# AVERAGE POPULARITY BY DEPARTMENT
# ==========================================

avg_pop = (
    df.groupby('known_for_department')['popularity']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    avg_pop,
    x='known_for_department',
    y='popularity',
    color='popularity',
    title='Average Popularity by Department',
    template='plotly_dark'
)

fig.show()
```


# 📊 Advanced Multi-Visualization Dashboard


```

# ==========================================
# DASHBOARD
# ==========================================

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Gender Distribution',
        'Department Distribution',
        'Popularity Distribution',
        'Average Popularity'
    )
)

# Gender
gender_counts = df['gender_label'].value_counts()

fig.add_trace(
    go.Pie(
        labels=gender_counts.index,
        values=gender_counts.values
    ),
    row=1,
    col=1
)

# Department
dept = df['known_for_department'].value_counts()

fig.add_trace(
    go.Bar(
        x=dept.index,
        y=dept.values
    ),
    row=1,
    col=2
)

# Popularity
fig.add_trace(
    go.Histogram(
        x=df['popularity']
    ),
    row=2,
    col=1
)

# Average popularity
fig.add_trace(
    go.Bar(
        x=avg_pop['known_for_department'],
        y=avg_pop['popularity']
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=900,
    title='Interactive Celebrity Analytics Dashboard',
    template='plotly_dark'
)

fig.show()
```
