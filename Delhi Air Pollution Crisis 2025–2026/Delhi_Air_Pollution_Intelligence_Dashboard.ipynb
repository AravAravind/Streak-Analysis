```

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

df = pd.read_csv(
    '/kaggle/input/delhi-air-quality-time-series/Delhi_Air_Quality_Time_Series_Dataset_01-01-2025_to_15-05-2026_.csv'
)

print(df.shape)

df.head()
```

```

df['DateTime'] = pd.to_datetime(df['DateTime'])

df.loc[df['Values'] < -100, 'Values'] = np.nan

df.info()
```


# 🌫️ Pollution Distribution


```

fig = px.histogram(
    df,
    x='Values',
    nbins=80,
    color='Parameters',
    template='plotly_dark',
    title='Pollution Value Distribution'
)

fig.show(renderer='iframe_connected')
```


# 📍 Station-wise Pollution Intelligence


```

station_avg = (
    df.groupby('Locations')['Values']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    station_avg,
    x='Locations',
    y='Values',
    color='Values',
    template='plotly_dark',
    title='Average Pollution by Station'
)

fig.show(renderer='iframe_connected')
```


# ⏰ Hourly Pollution Trend


```

hourly = (
    df.groupby(['hour', 'Parameters'])['Values']
    .mean()
    .reset_index()
)

fig = px.line(
    hourly,
    x='hour',
    y='Values',
    color='Parameters',
    template='plotly_dark',
    title='Hourly Pollution Trend'
)

fig.show(renderer='iframe_connected')
```


# 🔥 Pollutant Correlation Heatmap


```

pivot_corr = (
    df.pivot_table(
        index='DateTime',
        columns='Parameters',
        values='Values',
        aggfunc='mean'
    )
)

corr = pivot_corr.corr()

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Pollutant Correlation Heatmap')

plt.show()
```


# 🧠 PM2.5 Crisis Analysis


```

pm25 = df[df['Parameters'] == 'pm25']

daily_pm25 = (
    pm25.groupby(pm25['DateTime'].dt.date)['Values']
    .mean()
    .reset_index()
)

fig = px.line(
    daily_pm25,
    x='DateTime',
    y='Values',
    template='plotly_dark',
    title='Daily PM2.5 Trend'
)

fig.add_hline(
    y=45,
    line_dash='dash',
    annotation_text='WHO Guideline'
)

fig.show(renderer='iframe_connected')
```


# 🚀 Executive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Station Pollution',
        'Hourly Trend',
        'Pollutant Comparison'
    )
)

# Distribution
fig.add_trace(
    go.Histogram(
        x=df['Values']
    ),
    row=1,
    col=1
)

# Station
station = (
    df.groupby('Locations')['Values']
    .mean()
)

fig.add_trace(
    go.Bar(
        x=station.index,
        y=station.values
    ),
    row=1,
    col=2
)

# Hourly
hourly2 = (
    df.groupby('hour')['Values']
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=hourly2.index,
        y=hourly2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)

# Pollutants
pollutants = (
    df.groupby('Parameters')['Values']
    .mean()
)

fig.add_trace(
    go.Bar(
        x=pollutants.index,
        y=pollutants.values
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=950,
    template='plotly_dark',
    title='🌫️ Delhi Air Quality Intelligence Dashboard',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```


# 🧠 Key Insights

- PM2.5 levels frequently exceed WHO guidelines
- Pollution intensity varies strongly by station
- Hourly trends reveal dangerous evening spikes
- Weather and pollutants are highly interconnected
