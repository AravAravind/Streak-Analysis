```

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default='notebook_connected'

sns.set_style('whitegrid')
```

```

df = pd.read_csv('/kaggle/input/athlete-data/athlete_data.csv')

print(df.shape)
df.head()
```

## 📊 Dataset Overview

```

df.info()
```

```

df.describe()
```

## 🚑 Injury Risk Distribution

```

inj = df['Injury_Indicator'].value_counts()

fig = px.pie(
    values=inj.values,
    names=inj.index,
    hole=0.5,
    title='Athlete Injury Distribution'
)

fig.show()
```

## 🔥 What Drives Injury Risk?

```

risk = (
    df.groupby('Training_Intensity')['ACL_Risk_Score']
    .mean()
    .reset_index()
)

fig = px.bar(
    risk,
    x='Training_Intensity',
    y='ACL_Risk_Score',
    color='ACL_Risk_Score',
    title='Average ACL Risk by Training Intensity'
)

fig.show()
```

## ⚡ Fatigue vs Performance

```

fig = px.scatter(
    df,
    x='Fatigue_Score',
    y='Performance_Score',
    color='Injury_Indicator',
    size='Training_Hours_Per_Week',
    title='Fatigue vs Performance'
)

fig.show()
```

## 🧠 Recovery Intelligence

```

recovery = (
    df.groupby('Recovery_Days_Per_Week')
    [['Performance_Score','ACL_Risk_Score']]
    .mean()
    .reset_index()
)

fig = px.line(
    recovery,
    x='Recovery_Days_Per_Week',
    y=['Performance_Score','ACL_Risk_Score'],
    markers=True,
    title='Recovery Impact'
)

fig.show()
```

## ❤️ Heart Rate & Performance

```

fig = px.scatter(
    df,
    x='heartbeat',
    y='Performance_Score',
    color='Training_Intensity',
    title='Heart Rate vs Performance'
)

fig.show()
```

## 🏋️ Training Load Analysis

```

fig = px.box(
    df,
    x='Training_Intensity',
    y='Performance_Score',
    color='Training_Intensity',
    title='Performance by Training Intensity'
)

fig.show()
```

## 🔥 Correlation Heatmap

```

num = df.select_dtypes(include='number')

plt.figure(figsize=(14,8))

sns.heatmap(
    num.corr(),
    cmap='coolwarm',
    annot=True
)

plt.title('Sports Science Correlation Matrix')
plt.show()
```

## 🚀 Executive Sports Dashboard

```

fig = make_subplots(
rows=2,
cols=2,
subplot_titles=(
'Fatigue Distribution',
'Performance Distribution',
'Training Load',
'ACL Risk'
)
)

fig.add_trace(
go.Histogram(x=df['Fatigue_Score']),
row=1,col=1
)

fig.add_trace(
go.Histogram(x=df['Performance_Score']),
row=1,col=2
)

load=df.groupby('Training_Intensity')['Performance_Score'].mean()

fig.add_trace(
go.Bar(
x=load.index,
y=load.values
),
row=2,col=1
)

risk=df.groupby('Training_Intensity')['ACL_Risk_Score'].mean()

fig.add_trace(
go.Bar(
x=risk.index,
y=risk.values
),
row=2,col=2
)

fig.update_layout(
height=900,
title='🏆 Athlete Intelligence Dashboard',
showlegend=False
)

fig.show()
```


# 🎯 Storytelling Insights

### Insight 1
Higher training load does not always translate into higher performance.

### Insight 2
Recovery days can reduce injury exposure while maintaining performance.

### Insight 3
Fatigue and injury risk often move together.

### Insight 4
ACL risk can be profiled using workload and recovery variables.

### Insight 5
Sports analytics can identify athlete profiles requiring intervention.
