```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(r"/mnt/data/tech_layoffs_hiring_trends_elite_v2.csv")

print("Dataset Loaded Successfully!")
print(df.shape)

df.head()
```

```

# ==========================================
# DATASET INFO
# ==========================================

df.info()
```

```

# ==========================================
# MISSING VALUES
# ==========================================

missing = df.isnull().sum().sort_values(ascending=False)

missing.head(20)
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 📈 Exploratory Data Analysis


```

# ==========================================
# DISTRIBUTION ANALYSIS
# ==========================================

fig = px.histogram(
    df,
    x='year',
    nbins=60,
    title='Distribution of year',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```

```

# ==========================================
# CORRELATION HEATMAP
# ==========================================

corr = df[['year', 'layoffs_count', 'layoff_percentage', 'ai_automation_impact', 'ai_replacement_risk', 'open_roles', 'remote_jobs_percentage', 'stock_growth_percent', 'revenue_growth_percent', 'salary_budget_change']].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Tech Industry Correlation Heatmap')

plt.show()
```


# 🏢 Top Companies Analysis


```

top_company = (
    df.groupby('record_id')['year']
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_company,
    x='record_id',
    y='year',
    color='year',
    title='Top Companies Comparison',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 📊 Hiring & Layoff Trend Analysis


```

trend = (
    df.groupby('month')['year']
    .mean()
    .reset_index()
)

fig = px.line(
    trend,
    x='month',
    y='year',
    markers=True,
    title='Industry Trend Analysis',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🔍 Workforce Relationship Analysis


```

fig = px.scatter(
    df,
    x='year',
    y='layoffs_count',
    opacity=0.7,
    title='year vs layoffs_count',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🚀 Executive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Top Companies',
        'Trend Analysis',
        'Scatter Insights'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['year']
    ),
    row=1,
    col=1
)


top10 = (
    df.groupby('record_id')['year']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)


trend2 = (
    df.groupby('month')['year']
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=trend2.index,
        y=trend2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)


fig.add_trace(
    go.Scatter(
        x=df['year'],
        y=df['layoffs_count'],
        mode='markers'
    ),
    row=2,
    col=2
)


fig.update_layout(
    height=950,
    title='🚨 Tech Layoffs & Hiring Trends Dashboard',
    template='plotly',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
