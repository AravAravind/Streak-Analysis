```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(
    '/kaggle/input/mental-health-workplace/mental_health_workplace.csv'
)

print("Dataset Loaded Successfully!")

print(df.shape)

df.head()
```

```

# ==========================================
# DATASET INFO
# ==========================================

df.info()
```

```

# ==========================================
# MISSING VALUES
# ==========================================

missing = df.isnull().sum().sort_values(ascending=False)

missing.head(20)
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 👥 Age Distribution Analysis


```

fig = px.histogram(
    df,
    x='age_group',
    nbins=40,
    title='Employee Age Distribution',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🚻 Gender Distribution


```

gender_counts = df['gender'].value_counts()

fig = px.pie(
    values=gender_counts.values,
    names=gender_counts.index,
    hole=0.5,
    title='Gender Distribution',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 😰 Workplace Stress Analysis


```

fig = px.box(
    df,
    y='stress_level',
    title='Stress Level Distribution',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🔥 Mental Health Correlation Analysis


```

corr = df[['year', 'years_of_experience', 'annual_salary_usd', 'weekly_work_hours', 'weekly_overtime_hours', 'burnout_risk_score', 'work_life_balance_score', 'productivity_score', 'job_satisfaction_score', 'absenteeism_days_per_year']].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Mental Health Correlation Heatmap')

plt.show()
```


# 🔍 Psychological Relationship Analysis


```

fig = px.scatter(
    df,
    x='year',
    y='years_of_experience',
    opacity=0.7,
    title='year vs years_of_experience',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🚀 Employee Wellness Executive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    specs=[
        [{"type": "xy"}, {"type": "domain"}],
        [{"type": "xy"}, {"type": "xy"}]
    ],
    subplot_titles=(
        'Age Distribution',
        'Gender Distribution',
        'Stress Analysis',
        'Psychological Relationship'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['age_group']
    ),
    row=1,
    col=1
)


gender_counts = df['gender'].value_counts()

fig.add_trace(
    go.Pie(
        labels=gender_counts.index,
        values=gender_counts.values,
        hole=0.4
    ),
    row=1,
    col=2
)


fig.add_trace(
    go.Box(
        y=df['stress_level']
    ),
    row=2,
    col=1
)


fig.add_trace(
    go.Scatter(
        x=df['year'],
        y=df['years_of_experience'],
        mode='markers'
    ),
    row=2,
    col=2
)


fig.update_layout(
    height=950,
    title='🧠 Employee Wellness Intelligence Dashboard',
    template='plotly',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
