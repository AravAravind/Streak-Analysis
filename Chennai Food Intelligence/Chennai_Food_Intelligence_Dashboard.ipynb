```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASETS
# ==========================================

final_df = pd.read_csv(
    '/kaggle/input/zomato-chennai/Zomato_Chennai_Final.csv',
    encoding='latin1'
)

segmented_df = pd.read_csv(
    '/kaggle/input/zomato-chennai/Zomato_Chennai_Segmented.csv',
    encoding='latin1'
)

print("Final Dataset Shape:", final_df.shape)
print("Segmented Dataset Shape:", segmented_df.shape)
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

final_df.describe(include='all')
```


# ⭐ Restaurant Ratings Analysis


```

fig = px.histogram(
    final_df,
    x='dining_rating',
    nbins=40,
    title='Restaurant Ratings Distribution',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🍜 Cuisine Popularity Analysis


```

top_cuisines = (
    final_df['cuisine']
    .value_counts()
    .head(15)
)

fig = px.bar(
    x=top_cuisines.index,
    y=top_cuisines.values,
    color=top_cuisines.values,
    title='Top Cuisines in Chennai',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 📍 Chennai Area Intelligence


```

top_locations = (
    final_df.groupby('area/location')['name_of_restaurant']
    .count()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_locations,
    x='area/location',
    y='name_of_restaurant',
    color='name_of_restaurant',
    title='Top Restaurant Areas in Chennai',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🔥 Restaurant Correlation Analysis


```

corr = final_df[['latitude', 'longitude', 'dining_rating']].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Restaurant Intelligence Correlation Heatmap')

plt.show()
```


# 🧩 Restaurant Segmentation Analysis


```

segmented_df.info()
```

```

segment_counts = segmented_df['Name of Restaurant'].value_counts()

fig = px.pie(
    values=segment_counts.values,
    names=segment_counts.index,
    hole=0.5,
    title='Restaurant Segment Distribution',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🚀 Chennai Food Executive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    specs=[
        [{"type": "xy"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "domain"}]
    ],
    subplot_titles=(
        'Ratings Distribution',
        'Cuisine Popularity',
        'Cost vs Rating',
        'Restaurant Segments'
    )
)


fig.add_trace(
    go.Histogram(
        x=final_df['dining_rating']
    ),
    row=1,
    col=1
)


top10 = (
    final_df['cuisine']
    .value_counts()
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)


segment_counts = segmented_df['Name of Restaurant'].value_counts()

fig.add_trace(
    go.Pie(
        labels=segment_counts.index,
        values=segment_counts.values,
        hole=0.4
    ),
    row=2,
    col=2
)


fig.update_layout(
    height=950,
    title='🍽️ Chennai Food Intelligence Dashboard',
    template='plotly',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
