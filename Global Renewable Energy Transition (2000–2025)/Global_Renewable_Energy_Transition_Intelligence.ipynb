```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go

from plotly.subplots import make_subplots

import plotly.io as pio
import plotly.offline as pyo

# Kaggle-friendly rendering
pio.renderers.default = "notebook_connected"

pyo.init_notebook_mode()

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(
    '/kaggle/input/global-renewable-energy-transition-2000-2025/global_renewable_energy_transition_2000_2025.csv'
)

print("Dataset Loaded Successfully!")

print(df.shape)

df.head()
```

```

# ==========================================
# DATASET INFO
# ==========================================

df.info()
```

```

# ==========================================
# MISSING VALUES
# ==========================================

missing = df.isnull().sum().sort_values(ascending=False)

missing.head(20)
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 📈 Renewable Energy Distribution


```

fig = px.histogram(
    df,
    x='year',
    nbins=60,
    color_discrete_sequence=['green'],
    title='Distribution of year',
    template='plotly_dark'
)

fig.show()
```


# 🌍 Top Renewable Energy Leaders


```

top_country = (
    df.groupby('country')['renewables_electricity_twh']
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_country,
    x='country',
    y='renewables_electricity_twh',
    color='renewables_electricity_twh',
    title='Top Renewable Energy Leaders',
    template='plotly_dark'
)

fig.show()
```


# 📊 Renewable Energy Growth Trends


```

top5 = (
    df.groupby('country')['renewables_electricity_twh']
    .mean()
    .sort_values(ascending=False)
    .head(5)
    .index
)

trend = df[df['country'].isin(top5)]

fig = px.line(
    trend,
    x='year',
    y='renewables_electricity_twh',
    color='country',
    title='Top Countries Renewable Energy Trend',
    template='plotly_dark'
)

fig.show()
```


# 🗺️ Global Renewable Energy Map


```

country_avg = (
    df.groupby('country')['renewables_electricity_twh']
    .mean()
    .reset_index()
)

fig = px.choropleth(
    country_avg,
    locations='country',
    locationmode='country names',
    color='renewables_electricity_twh',
    hover_name='country',
    title='Global Renewable Energy Adoption',
    template='plotly_dark'
)

fig.show()
```


# 🔥 Sustainability Correlation Analysis


```

corr = df[['year', 'population', 'gdp_usd', 'total_electricity_generation_twh', 'electricity_demand_twh', 'solar_electricity_twh', 'wind_electricity_twh', 'renewables_electricity_twh', 'hydro_electricity_twh', 'nuclear_electricity_twh']].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Renewable Energy Correlation Heatmap')

plt.show()
```


# 🔍 Energy Relationship Analysis


```

fig = px.scatter(
    df,
    x='year',
    y='population',
    opacity=0.7,
    title='year vs population',
    template='plotly_dark'
)

fig.show()
```


# 🚀 Climate Intelligence Executive Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Top Countries',
        'Growth Trend',
        'Sustainability Scatter'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['year']
    ),
    row=1,
    col=1
)


top10 = (
    df.groupby('country')['renewables_electricity_twh']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)


trend2 = (
    df.groupby('year')['renewables_electricity_twh']
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=trend2.index,
        y=trend2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)


fig.add_trace(
    go.Scatter(
        x=df['year'],
        y=df['population'],
        mode='markers'
    ),
    row=2,
    col=2
)


fig.update_layout(
    height=950,
    title='🌍 Global Renewable Energy Intelligence Dashboard',
    template='plotly_dark',
    showlegend=False
)

fig.show()
```
