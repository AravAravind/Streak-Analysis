```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(
    '/kaggle/input/renewable-energy-share-2000-2025/renewable_energy_share_2000_2025.csv'
)

print("Dataset Loaded Successfully!")

print(df.shape)

df.head()
```

```

# ==========================================
# DATASET INFO
# ==========================================

df.info()
```

```

# ==========================================
# MISSING VALUES
# ==========================================

missing = df.isnull().sum().sort_values(ascending=False)

missing.head(20)
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 📈 Renewable Energy Distribution


```

fig = px.histogram(
    df,
    x='year',
    nbins=60,
    title='Distribution of year',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🔥 Correlation Analysis


```

corr = df[['year', 'population', 'gdp', 'primary_energy_consumption', 'electricity_generation', 'electricity_demand', 'renewables_share_energy', 'renewables_share_elec', 'fossil_share_energy', 'fossil_share_elec']].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Renewable Energy Correlation Heatmap')

plt.show()
```


# 🌍 Top Renewable Energy Leaders


```

top_country = (
    df.groupby('country')['year']
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_country,
    x='country',
    y='year',
    color='year',
    title='Top Renewable Energy Leaders',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 📊 Renewable Energy Growth Trends


```

top5 = (
    df.groupby('country')['year']
    .mean()
    .sort_values(ascending=False)
    .head(5)
    .index
)

trend = df[df['country'].isin(top5)]

fig = px.line(
    trend,
    x='year',
    y='year',
    color='country',
    title='Top Countries Renewable Energy Trend',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🗺️ Global Renewable Energy Map


```

country_avg = (
    df.groupby('country')['year']
    .mean()
    .reset_index()
)

fig = px.choropleth(
    country_avg,
    locations='country',
    locationmode='country names',
    color='year',
    hover_name='country',
    title='Global Renewable Energy Adoption',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🔍 Sustainability Relationship Analysis


```

fig = px.scatter(
    df,
    x='year',
    y='population',
    opacity=0.7,
    title='year vs population',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🚀 Climate Intelligence Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Top Countries',
        'Growth Trend',
        'Sustainability Scatter'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['year']
    ),
    row=1,
    col=1
)


top10 = (
    df.groupby('country')['year']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)


trend2 = (
    df.groupby('year')['year']
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=trend2.index,
        y=trend2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)


fig.add_trace(
    go.Scatter(
        x=df['year'],
        y=df['population'],
        mode='markers'
    ),
    row=2,
    col=2
)


fig.update_layout(
    height=950,
    title='🌱 Renewable Energy Intelligence Dashboard',
    template='plotly',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
