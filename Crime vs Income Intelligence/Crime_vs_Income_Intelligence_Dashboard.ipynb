```

# ==========================================
# IMPORT LIBRARIES
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

import warnings
warnings.filterwarnings("ignore")

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(
    '/kaggle/input/state-crime-income-merged/state_crime_income_merged.csv'
)

print("Dataset Loaded Successfully!")

print(df.shape)

df.head()
```

```

# ==========================================
# DATASET INFO
# ==========================================

df.info()
```

```

# ==========================================
# MISSING VALUES
# ==========================================

missing = df.isnull().sum().sort_values(ascending=False)

missing.head(20)
```

```

# ==========================================
# DESCRIPTIVE STATISTICS
# ==========================================

df.describe(include='all')
```


# 🚔 Crime Distribution Analysis


```

fig = px.histogram(
    df,
    x='Crime Count_aggravated_assault',
    nbins=50,
    title='Crime Distribution',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🏙️ State Crime Intelligence


```

top_states = (
    df.groupby('State')['Crime Count_aggravated_assault']
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_states,
    x='State',
    y='Crime Count_aggravated_assault',
    color='Crime Count_aggravated_assault',
    title='Highest Crime States',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 📈 Crime Trend Analysis


```

trend = (
    df.groupby('Year')['Crime Count_aggravated_assault']
    .mean()
    .reset_index()
)

fig = px.line(
    trend,
    x='Year',
    y='Crime Count_aggravated_assault',
    markers=True,
    title='Crime Trend Over Time',
    template='plotly'
)

fig.show(renderer='iframe_connected')
```


# 🔥 Correlation Intelligence


```

corr = df[['Year', 'Crime Count_aggravated_assault', 'Crime Count_burglary', 'Crime Count_homicide', 'Crime Count_larceny', 'Crime Count_motor_vehicle_theft', 'Crime Count_robbery', 'Crime Rate_aggravated_assault', 'Crime Rate_burglary', 'Crime Rate_homicide']].corr(numeric_only=True)

plt.figure(figsize=(14,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Socioeconomic Correlation Heatmap')

plt.show()
```


# 🚀 Executive Intelligence Dashboard


```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Crime Distribution',
        'Income Distribution',
        'Crime Trend',
        'Crime vs Income'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['Crime Count_aggravated_assault']
    ),
    row=1,
    col=1
)


trend2 = (
    df.groupby('Year')['Crime Count_aggravated_assault']
    .mean()
)

fig.add_trace(
    go.Scatter(
        x=trend2.index,
        y=trend2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)


fig.update_layout(
    height=950,
    title='🚨 Crime vs Income Intelligence Dashboard',
    template='plotly',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
