```

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

import plotly.io as pio
pio.renderers.default = "iframe_connected"

sns.set_style("darkgrid")

print("Libraries loaded successfully!")
```

```

df = pd.read_csv('/kaggle/input/netflix-dataset/01_Netflix_2016_2025.csv', encoding='latin1')

print(df.shape)

df.head()
```

```

df.info()
```

```

df.describe(include='all')
```

```

missing = df.isnull().sum().sort_values(ascending=False)

missing.head(20)
```

## 📈 Distribution Analysis

```

fig = px.histogram(
    df,
    x='Year',
    nbins=60,
    template='plotly',
    title='Distribution of Year'
)

fig.show(renderer='iframe_connected')
```

## 🍿 Genre Analysis

```

top_genres = (
    df['Genre']
    .value_counts()
    .head(15)
)

fig = px.bar(
    x=top_genres.index,
    y=top_genres.values,
    color=top_genres.values,
    template='plotly',
    title='Top Netflix Genres'
)

fig.show(renderer='iframe_connected')
```

## 📅 Release Trend Analysis

```

trend = (
    df.groupby('Year')
    .size()
    .reset_index(name='count')
)

fig = px.line(
    trend,
    x='Year',
    y='count',
    markers=True,
    template='plotly',
    title='Netflix Release Trend'
)

fig.show(renderer='iframe_connected')
```

## 🚀 Executive Dashboard

```

fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        'Distribution',
        'Genre Popularity',
        'Release Trend',
        'Ratings'
    )
)


fig.add_trace(
    go.Histogram(
        x=df['Year']
    ),
    row=1,
    col=1
)


top10 = (
    df['Genre']
    .value_counts()
    .head(10)
)

fig.add_trace(
    go.Bar(
        x=top10.index,
        y=top10.values
    ),
    row=1,
    col=2
)


trend2 = (
    df.groupby('Year')
    .size()
)

fig.add_trace(
    go.Scatter(
        x=trend2.index,
        y=trend2.values,
        mode='lines+markers'
    ),
    row=2,
    col=1
)


fig.update_layout(
    height=950,
    template='plotly',
    title='🎬 Netflix Streaming Intelligence Dashboard',
    showlegend=False
)

fig.show(renderer='iframe_connected')
```
