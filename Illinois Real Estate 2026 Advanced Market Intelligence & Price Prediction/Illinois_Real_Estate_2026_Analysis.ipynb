```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
```

```

# Load dataset
df = pd.read_csv('/mnt/data/Pasted text(12).txt')

df.head()
```

## 🧹 Data Cleaning

```

# Convert date
df['soldOn'] = pd.to_datetime(df['soldOn'], errors='coerce')

# Fill missing values
df = df.fillna(df.median(numeric_only=True))

# Feature Engineering
df['price_diff'] = df['listPrice'] - df['lastSoldPrice']
df['sale_year'] = df['soldOn'].dt.year

df.head()
```

## 📊 Visualizations

```

# Price Distribution
plt.figure()
df['lastSoldPrice'].hist(bins=20)
plt.title('Sold Price Distribution')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```

# Price vs Sqft
plt.figure()
plt.scatter(df['sqft'], df['lastSoldPrice'])
plt.title('Price vs Sqft')
plt.xlabel('Sqft')
plt.ylabel('Price')
plt.show()
```

## 🤖 Machine Learning Model

```

features = ['sqft','beds','baths','garage','year_built']
target = 'lastSoldPrice'

X = df[features]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestRegressor()
model.fit(X_train, y_train)

preds = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test, preds))
print("R2:", r2_score(y_test, preds))
```

```

importances = model.feature_importances_

plt.figure()
plt.bar(features, importances)
plt.title("Feature Importance")
plt.show()
```
